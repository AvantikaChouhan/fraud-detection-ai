```python
from pyspark.sql.functions import current_timestamp

source_df = spark.table(
    "workspace.default.fraud_explanations_source"
)

genai_df = source_df.withColumn(
    "genai_processed_time",
    current_timestamp()
)

(
    genai_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(
        "workspace.default.fraud_explanations"
    )
)

print("GenAI Delta table created successfully!")
print("Rows:", genai_df.count())
```
